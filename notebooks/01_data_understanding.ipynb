## Observations

* The data covers 36 months, January 2023 to December 2025, with six channel spends, eight external factors, revenue, units sold and three funnel counts.
* There are no missing values, no duplicate rows and no skipped months. Every column already has a suitable type except `month`, which is read as text and is converted to a date in notebook 02.
* Four text columns (`country`, `brand_category`, `region_scope`, `data_type`) have one value each, so they carry no information for modelling.
* `total_ad_spend_lkr` differs from the sum of the six channels by 1 or 2 LKR in 16 rows. This is rounding and is handled in notebook 02.
* `impressions`, `clicks` and `conversions` are results of advertising, not causes of sales, so they should not be used as predictors of revenue.
* Every row is marked SYNTHETIC in `data_type`, and the dictionary says the exchange rate and inflation values are not real history. Any conclusion applies to this simulated data only.
